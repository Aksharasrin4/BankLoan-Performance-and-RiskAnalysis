```python
# Adjust Date Columns to Handle Mixed Formats
import pandas as pd
from dateutil import parser

def parse_mixed_dates(date_series):
    """Parse a Series with mixed date formats, returning datetime."""
    def safe_parse(val):
        if pd.isna(val) or str(val).strip() == '':
            return pd.NaT
        try:
            return parser.parse(str(val).strip(), dayfirst=True)
        except Exception:
            return pd.NaT
    
    return date_series.apply(safe_parse)

# Define your date columns
date_cols = ['issue_date', 'last_credit_pull_date', 'last_payment_date', 'next_payment_date']

# Apply to each date column
for col in date_cols:
    df[col] = parse_mixed_dates(df[col])

# Optional: standardize output format
for col in date_cols:
    df[col] = df[col].dt.strftime('%m-%d-%Y')

df[date_cols].head()
```

```python
date_cols = ['issue_date', 'last_credit_pull_date', 'last_payment_date', 'next_payment_date']

for col in date_cols:
    df[col] = pd.to_datetime(df[col], format='%m-%d-%Y', errors='coerce')
```

```python
df['term'] = df['term'].str.strip()  # " 60 months" → "60 months"
```

```python
df['emp_length'] = df['emp_length'].str.strip()
# Check unique values
df['emp_length'].unique()
```

```python
#  Standardize 'emp_length' values
emp_length_map = {
    '< 1 year': 0,
    '1 year': 1,
    '2 years': 2,
    '3 years': 3,
    '4 years': 4,
    '5 years': 5,
    '6 years': 6,
    '7 years': 7,
    '8 years': 8,
    '9 years': 9,
    '10+ years': 10
}

df['emp_length'] = df['emp_length'].map(emp_length_map)
```

```python
# 'Debt consolidation' should be 'debt_consolidation', mixed styles overall
# Standardize to lowercase with underscores
df['purpose'] = df['purpose'].str.lower().str.replace(' ', '_')
```
